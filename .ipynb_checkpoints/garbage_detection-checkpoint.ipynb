```python
import json, os, shutil, glob, zipfile


from huggingface_hub import hf_hub_download
from ultralytics import YOLO
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
BASE = os.path.expanduser(
    "~/Garbage-Waste-Object-Detection-YOLOv8/data/garbage_raw"
)
os.makedirs(BASE, exist_ok=True)

REPO_ID = "keremberke/garbage-object-detection"

for split in ["train", "valid", "test"]:
    zip_path = hf_hub_download(repo_id=REPO_ID, repo_type="dataset", filename=f"data/{split}.zip")
    out_dir = f"{BASE}/{split}"
    if not os.path.exists(out_dir):
        with zipfile.ZipFile(zip_path) as z:
            z.extractall(out_dir)
    print(split, "->", out_dir, "(", len(os.listdir(out_dir)), "files )")
```

```python
def coco_to_yolo(split_dir, out_images_dir, out_labels_dir, class_list=None):
    json_candidates = glob.glob(os.path.join(split_dir, '**', '*.json'), recursive=True)
    assert json_candidates, f"No COCO json found under {split_dir}"
    json_path = json_candidates[0]
    images_root = os.path.dirname(json_path)  # images usually live next to the json

    with open(json_path) as f:
        coco = json.load(f)

    categories = sorted(coco['categories'], key=lambda c: c['id'])
    if class_list is None:
        class_list = [c['name'] for c in categories]
    name_by_id = {c['id']: c['name'] for c in categories}
    cat_id_to_idx = {cid: class_list.index(name) for cid, name in name_by_id.items() if name in class_list}

    images_by_id = {img['id']: img for img in coco['images']}
    anns_by_image = {}
    for ann in coco['annotations']:
        anns_by_image.setdefault(ann['image_id'], []).append(ann)

    os.makedirs(out_images_dir, exist_ok=True)
    os.makedirs(out_labels_dir, exist_ok=True)

    n_written = 0
    for img_id, img in images_by_id.items():
        fname = img['file_name']
        src_path = os.path.join(images_root, fname)
        if not os.path.exists(src_path):
            continue
        shutil.copy(src_path, os.path.join(out_images_dir, fname))

        w, h = img['width'], img['height']
        lines = []
        for ann in anns_by_image.get(img_id, []):
            x, y, bw, bh = ann['bbox']
            cx, cy = (x + bw / 2) / w, (y + bh / 2) / h
            nw, nh = bw / w, bh / h
            cls_idx = cat_id_to_idx[ann['category_id']]
            lines.append(f"{cls_idx} {cx:.6f} {cy:.6f} {nw:.6f} {nh:.6f}")

        label_name = os.path.splitext(fname)[0] + '.txt'
        with open(os.path.join(out_labels_dir, label_name), 'w') as f:
            f.write('\n'.join(lines))
        n_written += 1

    print(f"{split_dir}: wrote {n_written} images/labels -> classes={class_list}")
    return class_list
```

```python
YOLO_DIR = os.path.expanduser(
    "~/Garbage-Waste-Object-Detection-YOLOv8/data/garbage_yolo"
)

classes = coco_to_yolo(f"{BASE}/train", f"{YOLO_DIR}/train/images", f"{YOLO_DIR}/train/labels")
coco_to_yolo(f"{BASE}/valid", f"{YOLO_DIR}/valid/images", f"{YOLO_DIR}/valid/labels", class_list=classes)

data_yaml = f"""path: {YOLO_DIR}
train: train/images
val: valid/images
names: {classes}
"""
with open(f"{YOLO_DIR}/data.yaml", "w") as f:
    f.write(data_yaml)

print(data_yaml)
```

```python

RUNS_DIR = os.path.expanduser(
    "~/Garbage-Waste-Object-Detection-YOLOv8/runs"
)

model = YOLO('yolov8n.pt')

model.train(
    data=f"{YOLO_DIR}/data.yaml",
    epochs=25,
    imgsz=416,
    batch=16,
    patience=8,       # early stop if val metric plateaus
    project=RUNS_DIR,
    name='garbage_yolov8n',
    exist_ok=True,
)

TRAIN_DIR = model.trainer.save_dir
print("Training outputs saved to:", TRAIN_DIR)
```

```python
metrics = model.val(data=f"{YOLO_DIR}/data.yaml", project=RUNS_DIR, name='garbage_yolov8n_val', exist_ok=True)
VAL_DIR = metrics.save_dir

per_class_map50 = {
    classes[idx]: float(ap) for idx, ap in zip(metrics.box.ap_class_index, metrics.box.ap50)
}

print("mAP50:", metrics.box.map50)
print("mAP50-95:", metrics.box.map)
print("Per-class mAP50:", per_class_map50)
if len(per_class_map50) < len(classes):
    missing = set(classes) - set(per_class_map50)
    print("(no validation instances for:", missing, "- can't compute AP for these)")
print("Validation outputs saved to:", VAL_DIR)
```

```python
cm_path = os.path.join(str(VAL_DIR), 'confusion_matrix.png')
if os.path.exists(cm_path):
    plt.figure(figsize=(8,8))
    plt.imshow(mpimg.imread(cm_path))
    plt.axis('off')
    plt.show()
else:
    print("Not found at", cm_path, "- contents of VAL_DIR:", os.listdir(VAL_DIR))
```

```python
sample_images = glob.glob(f"{YOLO_DIR}/valid/images/*")[:6]
pred_results = model.predict(sample_images, conf=0.25, save=True, project=RUNS_DIR, name='garbage_preds', exist_ok=True)

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
for ax, r in zip(axes.flatten(), pred_results):
    ax.imshow(r.plot()[:, :, ::-1])  # BGR -> RGB
    ax.axis('off')
plt.tight_layout()
plt.show()

print("Prediction images saved to:", pred_results[0].save_dir)
```

```python
from pathlib import Path
from collections import defaultdict, Counter
import os
import yaml

YOLO_DIR = Path(
    os.path.expanduser(
        "~/Garbage-Waste-Object-Detection-YOLOv8/data/garbage_yolo"
    )
)

# Load class names
with open(YOLO_DIR / "data.yaml", "r") as f:
    data = yaml.safe_load(f)

classes = data["names"]

valid_labels = YOLO_DIR / "valid" / "labels"
valid_images = YOLO_DIR / "valid" / "images"

# ---------------------------------------------------------
# 1. Find images containing each class
# ---------------------------------------------------------

class_images = defaultdict(list)

for label_file in valid_labels.glob("*.txt"):

    present_classes = set()

    with open(label_file, "r") as f:
        for line in f:
            line = line.strip()

            if not line:
                continue

            class_id = int(line.split()[0])
            present_classes.add(class_id)

    for class_id in present_classes:
        class_images[class_id].append(label_file.stem)

print("=" * 75)
print("VALIDATION CLASS DISTRIBUTION")
print("=" * 75)

for class_id, class_name in enumerate(classes):

    images = class_images[class_id]

    print(f"\n{class_name.upper()}")
    print(f"Images containing class: {len(images)}")

    print("Example images:")
    for name in images[:10]:
        print("   ", name)


# ---------------------------------------------------------
# 2. Check how classes are distributed together
# ---------------------------------------------------------

print("\n" + "=" * 75)
print("CLASS COMBINATIONS IN VALIDATION SET")
print("=" * 75)

combinations = Counter()

for label_file in valid_labels.glob("*.txt"):

    present_classes = set()

    with open(label_file, "r") as f:
        for line in f:
            line = line.strip()

            if not line:
                continue

            class_id = int(line.split()[0])
            present_classes.add(class_id)

    combination = tuple(
        sorted(classes[c] for c in present_classes)
    )

    combinations[combination] += 1


for combination, count in combinations.most_common():

    print(f"{count:>5} images : {', '.join(combination)}")


# ---------------------------------------------------------
# 3. Check for empty label files
# ---------------------------------------------------------

empty_labels = []

for label_file in valid_labels.glob("*.txt"):

    if label_file.stat().st_size == 0:
        empty_labels.append(label_file.name)

print("\n" + "=" * 75)
print("EMPTY LABEL FILES")
print("=" * 75)

print("Number of empty label files:", len(empty_labels))

if empty_labels:
    print("Examples:")
    for name in empty_labels[:20]:
        print("   ", name)
```

```python
from collections import Counter
import glob
import os

classes = data["names"]

def count_classes(label_dir):
    counts = Counter()

    label_files = glob.glob(os.path.join(label_dir, "*.txt"))

    for label_file in label_files:
        with open(label_file, "r") as f:
            for line in f:
                line = line.strip()

                if not line:
                    continue

                class_id = int(line.split()[0])
                counts[class_id] += 1

    return counts


train_labels = os.path.join(YOLO_DIR, "train", "labels")
valid_labels = os.path.join(YOLO_DIR, "valid", "labels")

train_counts = count_classes(train_labels)
valid_counts = count_classes(valid_labels)

print("ACTUAL CLASSES FOUND IN LABEL FILES")
print("=" * 70)

for class_id, class_name in enumerate(classes):
    print(
        f"{class_id}: {class_name:<20} "
        f"Train objects = {train_counts[class_id]:>6}   "
        f"Valid objects = {valid_counts[class_id]:>6}"
    )
```

```python
all_ids = set(train_counts.keys()) | set(valid_counts.keys())

print("Class IDs found in labels:", sorted(all_ids))
print("Class IDs defined in YAML:", list(range(len(classes))))

unexpected = all_ids - set(range(len(classes)))

if unexpected:
    print(" Unexpected class IDs:", unexpected)
else:
    print("No unexpected class IDs found.")
```

```python
print("TACO CLASSES")
print("=" * 60)

for category in taco_data["categories"]:
    print(f"{category['id']:>3} : {category['name']}")
```

```python
category_ids = {
    category["id"]
    for category in taco_data["categories"]
}

annotation_category_ids = {
    annotation["category_id"]
    for annotation in taco_data["annotations"]
}

unknown_category_ids = annotation_category_ids - category_ids

print("Category IDs:", len(category_ids))
print("Used category IDs:", len(annotation_category_ids))
print("Unknown category IDs:", unknown_category_ids)
```

```python
# ============================================================
# TACO: Object count per class
# ============================================================

from collections import Counter

# Map category ID -> category name
category_map = {
    category["id"]: category["name"]
    for category in taco_data["categories"]
}

# Count annotations/objects per class
object_counts = Counter(
    annotation["category_id"]
    for annotation in taco_data["annotations"]
)

# Build inventory
taco_inventory = []

for category_id, category_name in category_map.items():
    taco_inventory.append({
        "category_id": category_id,
        "class": category_name,
        "objects": object_counts.get(category_id, 0)
    })

# Sort by number of objects
taco_inventory = sorted(
    taco_inventory,
    key=lambda x: x["objects"],
    reverse=True
)

for item in taco_inventory:
    print(
        f"{item['category_id']:>2} | "
        f"{item['class']:<30} | "
        f"{item['objects']:>4}"
    )
```

```python
# ============================================================
# TACO: Number of unique images containing each class
# ============================================================

images_per_class = Counter()

for annotation in taco_data["annotations"]:
    category_id = annotation["category_id"]
    images_per_class[category_id] += 1

# Build image-count inventory
for item in taco_inventory:
    category_id = item["category_id"]
    item["images"] = images_per_class.get(category_id, 0)

# Sort by image count
taco_inventory_by_images = sorted(
    taco_inventory,
    key=lambda x: x["images"],
    reverse=True
)

print(f"{'Class':<30} {'Images':>8} {'Objects':>8}")
print("=" * 50)

for item in taco_inventory_by_images:
    print(
        f"{item['class']:<30} "
        f"{item['images']:>8} "
        f"{item['objects']:>8}"
    )
```

```python
# ============================================================
# CURRENT DATASET INVENTORY
# Count images and objects for each class across all splits
# ============================================================

from pathlib import Path
from collections import defaultdict
import pandas as pd

CURRENT_DATASET_DIR = Path(
    os.path.expanduser(
        "~/Garbage-Waste-Object-Detection-YOLOv8/data/garbage_yolo"
    )
)

class_names = [
    "biodegradable",
    "cardboard",
    "glass",
    "metal",
    "paper",
    "plastic"
]

class_objects = defaultdict(int)
class_images = defaultdict(set)

for split in ["train", "valid", "test"]:

    label_dir = CURRENT_DATASET_DIR / split / "labels"

    if not label_dir.exists():
        print(f"Skipping {split}: labels directory not found")
        continue

    for label_file in label_dir.glob("*.txt"):

        image_id = label_file.stem

        with open(label_file, "r") as f:
            lines = [line.strip() for line in f if line.strip()]

        classes_in_image = set()

        for line in lines:
            parts = line.split()

            if len(parts) < 5:
                continue

            class_id = int(parts[0])

            if 0 <= class_id < len(class_names):
                class_name = class_names[class_id]

                class_objects[class_name] += 1
                classes_in_image.add(class_name)

        for class_name in classes_in_image:
            class_images[class_name].add(image_id)

current_inventory = pd.DataFrame({
    "class": class_names,
    "images": [
        len(class_images[c])
        for c in class_names
    ],
    "objects": [
        class_objects[c]
        for c in class_names
    ]
})

current_inventory = current_inventory.sort_values(
    "objects",
    ascending=False
).reset_index(drop=True)

current_inventory
```

```python
from pathlib import Path
from collections import Counter
import pandas as pd
import os

CURRENT_DATASET_DIR = Path(
    os.path.expanduser(
        "~/Garbage-Waste-Object-Detection-YOLOv8/data/garbage_yolo"
    )
)

class_names = [
    "biodegradable",
    "cardboard",
    "glass",
    "metal",
    "paper",
    "plastic"
]

rows = []

for split in ["train", "valid", "test"]:
    label_dir = CURRENT_DATASET_DIR / split / "labels"

    if not label_dir.exists():
        print(f"Skipping {split}: labels directory not found")
        continue

    for label_file in label_dir.glob("*.txt"):

        with open(label_file, "r") as f:
            lines = [line.strip() for line in f if line.strip()]

        class_counts = Counter()

        for line in lines:
            parts = line.split()

            if len(parts) < 5:
                continue

            class_id = int(parts[0])

            if 0 <= class_id < len(class_names):
                class_counts[class_names[class_id]] += 1

        for class_name, count in class_counts.items():
            rows.append({
                "split": split,
                "image": label_file.stem,
                "class": class_name,
                "objects_in_image": count
            })

current_object_distribution = pd.DataFrame(rows)

print(current_object_distribution.head())
```

```python
print("\nAverage objects per image by class:\n")

summary = (
    current_object_distribution
    .groupby("class")["objects_in_image"]
    .agg(
        images="count",
        total_objects="sum",
        average_objects_per_image="mean",
        max_objects_in_one_image="max"
    )
    .sort_values("total_objects", ascending=False)
)

summary
```

```python
from pathlib import Path
import os
import random
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image

CURRENT_DATASET_DIR = Path(
    os.path.expanduser(
        "~/Garbage-Waste-Object-Detection-YOLOv8/data/garbage_yolo"
    )
)

class_names = [
    "biodegradable",
    "cardboard",
    "glass",
    "metal",
    "paper",
    "plastic"
]

# Pick a few images containing each class
samples_per_class = 3

for class_id, class_name in enumerate(class_names):

    candidates = []

    for split in ["train", "valid"]:
        image_dir = CURRENT_DATASET_DIR / split / "images"
        label_dir = CURRENT_DATASET_DIR / split / "labels"

        if not label_dir.exists():
            continue

        for label_file in label_dir.glob("*.txt"):

            with open(label_file, "r") as f:
                lines = [line.strip() for line in f if line.strip()]

            contains_class = False

            for line in lines:
                parts = line.split()

                if len(parts) >= 5 and int(parts[0]) == class_id:
                    contains_class = True
                    break

            if contains_class:
                candidates.append(
                    (split, label_file)
                )

    random.shuffle(candidates)

    print(f"\n{class_name}: {len(candidates)} candidate images")

    for split, label_file in candidates[:samples_per_class]:

        # Find image
        possible_images = list(
            (CURRENT_DATASET_DIR / split / "images").glob(
                label_file.stem + ".*"
            )
        )

        if not possible_images:
            continue

        image_path = possible_images[0]

        img = Image.open(image_path).convert("RGB")

        fig, ax = plt.subplots(figsize=(10, 8))
        ax.imshow(img)

        img_w, img_h = img.size

        with open(label_file, "r") as f:
            lines = [line.strip() for line in f if line.strip()]

        for line in lines:

            parts = line.split()

            if len(parts) < 5:
                continue

            cid = int(parts[0])

            if cid != class_id:
                continue

            x_center, y_center, w, h = map(float, parts[1:5])

            x = (x_center - w / 2) * img_w
            y = (y_center - h / 2) * img_h
            box_w = w * img_w
            box_h = h * img_h

            rect = patches.Rectangle(
                (x, y),
                box_w,
                box_h,
                linewidth=2,
                edgecolor="red",
                facecolor="none"
            )

            ax.add_patch(rect)

            ax.text(
                x,
                y,
                class_name,
                fontsize=10,
                backgroundcolor="white"
            )

        ax.set_title(
            f"{class_name} | {split} | {label_file.stem}"
        )
        ax.axis("off")
        plt.show()
```

```python
# ============================================================
# DOWNLOAD TACO IMAGES USING annotations.json
# ============================================================

import json
import requests
from pathlib import Path
from PIL import Image
from io import BytesIO
from tqdm.auto import tqdm
import time

# Paths
TACO_ROOT = PROJECT_ROOT / "data" / "taco_raw" / "TACO"
TACO_DATA = TACO_ROOT / "data"
TACO_IMAGES = TACO_ROOT / "images"

TACO_IMAGES.mkdir(parents=True, exist_ok=True)

# Load annotations
with open(TACO_DATA / "annotations.json", "r") as f:
    taco_data = json.load(f)

images = taco_data["images"]

print("Images in annotations:", len(images))
print("Image directory:", TACO_IMAGES)
```

```python
# ============================================================
# DOWNLOAD MISSING TACO IMAGES
# ============================================================

downloaded = 0
already_exists = 0
failed = []

session = requests.Session()

for i, img_info in enumerate(tqdm(images, desc="Downloading TACO")):

    image_id = img_info["id"]
    filename = Path(img_info["file_name"]).name

    # Save using image ID so filenames are predictable
    output_path = TACO_IMAGES / f"{image_id:06d}.jpg"

    # Already downloaded
    if output_path.exists():
        already_exists += 1
        continue

    # Get URL
    url = img_info.get("flickr_640_url")

    if not url:
        url = img_info.get("flickr_url")

    if not url:
        failed.append((image_id, filename, "No URL"))
        continue

    try:

        response = session.get(
            url,
            timeout=30
        )

        response.raise_for_status()

        # Verify that response is actually an image
        img = Image.open(BytesIO(response.content)).convert("RGB")

        img.save(
            output_path,
            format="JPEG",
            quality=95
        )

        downloaded += 1

    except Exception as e:

        failed.append(
            (
                image_id,
                filename,
                str(e)
            )
        )

    # Small delay to avoid hammering Flickr
    time.sleep(0.05)

print("\n===================================")
print("TACO DOWNLOAD RESULT")
print("===================================")
print("Total annotation images :", len(images))
print("Downloaded              :", downloaded)
print("Already existed         :", already_exists)
print("Failed                  :", len(failed))
print("===================================")
```

```python
import json
import random
from pathlib import Path

from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/home/kumari-shikha/Garbage-Waste-Object-Detection-YOLOv8"
)

TACO_ROOT = PROJECT_ROOT / "data" / "taco_raw" / "TACO"
TACO_DATA = TACO_ROOT / "data"
TACO_IMAGES = TACO_ROOT / "images"

# ------------------------------------------------------------
# Load annotations
# ------------------------------------------------------------

with open(TACO_DATA / "annotations.json", "r") as f:
    taco_data = json.load(f)

# ------------------------------------------------------------
# Lookups
# ------------------------------------------------------------

category_name_to_id = {
    c["name"]: c["id"]
    for c in taco_data["categories"]
}

image_lookup = {
    img["id"]: img
    for img in taco_data["images"]
}

# ------------------------------------------------------------
# Group annotations by category
# ------------------------------------------------------------

annotations_by_category = {}

for ann in taco_data["annotations"]:

    category_id = ann["category_id"]

    annotations_by_category.setdefault(
        category_id,
        []
    ).append(ann)

# ------------------------------------------------------------
# Classes we want to inspect
# ------------------------------------------------------------

TACO_CANDIDATES = [
    "Drink can",
    "Clear plastic bottle",
    "Other plastic bottle",
    "Glass bottle",
    "Paper cup",
    "Disposable plastic cup",
    "Corrugated carton",
    "Other carton",
    "Normal paper",
    "Food waste",
]

# ------------------------------------------------------------
# Visual inspection
# ------------------------------------------------------------

random.seed(42)

for class_name in TACO_CANDIDATES:

    category_id = category_name_to_id[class_name]

    anns = annotations_by_category.get(
        category_id,
        []
    )

    image_ids = list(
        set(
            ann["image_id"]
            for ann in anns
        )
    )

    random.shuffle(image_ids)

    selected_ids = image_ids[:3]

    print(
        f"{class_name}: "
        f"{len(anns)} objects in "
        f"{len(image_ids)} images"
    )

    fig, axes = plt.subplots(
        1,
        len(selected_ids),
        figsize=(15, 5)
    )

    if len(selected_ids) == 1:
        axes = [axes]

    for ax, image_id in zip(
        axes,
        selected_ids
    ):

        image_path = (
            TACO_IMAGES /
            f"{image_id:06d}.jpg"
        )

        if not image_path.exists():

            ax.text(
                0.5,
                0.5,
                "Image not found",
                ha="center",
                va="center"
            )

            ax.axis("off")
            continue

        img = Image.open(
            image_path
        ).convert("RGB")

        ax.imshow(img)

        # Selected class annotations
        image_anns = [
            ann
            for ann in anns
            if ann["image_id"] == image_id
        ]

        for ann in image_anns:

            x, y, w, h = ann["bbox"]

            rect = patches.Rectangle(
                (x, y),
                w,
                h,
                linewidth=2,
                edgecolor="red",
                facecolor="none"
            )

            ax.add_patch(rect)

            ax.text(
                x,
                max(0, y - 5),
                class_name,
                fontsize=9,
                backgroundcolor="white"
            )

        ax.set_title(
            f"Image ID: {image_id}",
            fontsize=9
        )

        ax.axis("off")

    plt.suptitle(
        f"TACO — {class_name}",
        fontsize=15
    )

    plt.tight_layout()
    plt.show()
```
